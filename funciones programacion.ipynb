## 1. ¿Qué es una función?

Una **función** es un bloque de código con nombre que:
- Realiza una tarea específica
- Puede recibir datos de entrada (**parámetros**)
- Puede devolver un resultado (**return**)
- Se puede reutilizar cuantas veces quieras

**Analogía:** es como una licuadora 🥤  
Metés frutas (parámetros) → la licuadora hace su trabajo → salís con jugo (return).

## 2. Mi primera función

La sintaxis básica es:

```python
def nombre_funcion(parámetros):
    # cuerpo de la función
    return resultado
```

- `def` → le dice a Python que viene una definición de función
- `nombre_funcion` → el nombre que vos le dás
- `parámetros` → datos que la función necesita (pueden ser cero)
- `return` → valor que la función devuelve

````python
# ✏️ EJEMPLO 1 — Función sin parámetros
def saludar():
    print("¡Hola, futuro programador!")

# Para usar la función la llamamos por su nombre:
saludar()
saludar()   # Podemos llamarla múltiples veces
````

````python
# ✏️ EJEMPLO 2 — Función con parámetro
def saludar_persona(nombre):
    print(f"¡Hola, {nombre}! Bienvenido/a.")

saludar_persona("Ana")
saludar_persona("Carlos")
saludar_persona("Lucía")
````

## 3. Parámetros y Argumentos

| Término | Dónde aparece | Ejemplo |
|---------|---------------|---------|
| **Parámetro** | En la **definición** (`def`) | `def suma(a, b):` |
| **Argumento** | En la **llamada** | `suma(3, 5)` |

Los parámetros son como "variables reservadas" que esperan recibir datos cuando se llame la función.

````python
# ✏️ Función con múltiples parámetros
def calcular_area_rectangulo(base, altura):
    area = base * altura
    return area

# Distintos argumentos → distintos resultados
print(calcular_area_rectangulo(5, 3))   # 15
print(calcular_area_rectangulo(10, 4))  # 40
print(calcular_area_rectangulo(7, 7))   # 49
````

````python
# ✏️ Parámetros con valor por defecto
def saludar_con_idioma(nombre, idioma="español"):
    if idioma == "español":
        print(f"¡Hola, {nombre}!")
    elif idioma == "inglés":
        print(f"Hello, {nombre}!")
    else:
        print(f"Bonjour, {nombre}!")

saludar_con_idioma("Ana")                  # usa el default: español
saludar_con_idioma("John", "inglés")       # overridea el default
saludar_con_idioma("Marie", "francés")     # otro idioma
````

## 4. El `return`: devolver resultados

- `return` termina la función y devuelve un valor
- Sin `return`, la función devuelve `None`
- Podés usar el valor retornado en otras partes del código

````python
# ✏️ Con return — el valor queda disponible
def doblar(n):
    return n * 2

resultado = doblar(5)
print(resultado)               # 10
print(doblar(resultado) + 1)   # 21 — combinamos llamadas
print(type(doblar(3)))         # <class 'int'>
````

````python
# ✏️ Sin return — la función imprime pero no devuelve nada útil
def mostrar_doblar(n):
    print(f"El doble de {n} es {n * 2}")
    # No hay return → Python retorna None automáticamente

x = mostrar_doblar(5)    # Imprime pero x queda en None
print(f"x vale: {x}")   # None
````

````python
# ✏️ Python permite retornar múltiples valores
def minimo_y_maximo(lista):
    return min(lista), max(lista)   # devuelve una tupla

numeros = [5, 2, 8, 1, 9, 3]
menor, mayor = minimo_y_maximo(numeros)
print(f"Mínimo: {menor}, Máximo: {mayor}")
````

````python
# ✏️ Return temprano — útil para validaciones
def dividir(a, b):
    if b == 0:
        return "Error: no se puede dividir por cero"
    return a / b

print(dividir(10, 2))    # 5.0
print(dividir(10, 0))    # Error...
````

## 5. Scope (Ámbito de variables)

Las variables tienen un **alcance** (scope) que define desde dónde se pueden usar:

| Tipo | Dónde se define | Dónde se puede usar |
|------|-----------------|---------------------|
| **Global** | Fuera de funciones | En todo el programa |
| **Local** | Dentro de una función | Solo dentro de esa función |

````python
# ✏️ Variables globales vs locales
mensaje_global = "Soy global"    # Variable global

def mi_funcion():
    mensaje_local = "Soy local"  # Variable local
    print(mensaje_global)        # ✅ Puede acceder a la global
    print(mensaje_local)         # ✅ Puede acceder a la local

mi_funcion()
print(mensaje_global)            # ✅ Funciona
````

````python
# ✏️ Intentar acceder a una variable local desde afuera
def crear_variable():
    variable_secreta = 42

crear_variable()

# Descomentá la siguiente línea para ver el error:
# print(variable_secreta)   # NameError: name 'variable_secreta' is not defined
print("Si llegás aquí, no hubo error (la línea estaba comentada).")
````

````python
# ✏️ Buena práctica: usar return para "sacar" datos de la función
def calcular_doble(n):
    resultado = n * 2       # local
    return resultado        # lo sacamos con return

valor = calcular_doble(10)  # ahora 'valor' tiene 20
print(valor)
````

## 6. Funciones Built-in (predefinidas)

Python trae muchas funciones ya hechas. ¡No necesitás reinventar la rueda!

````python
# ✏️ Las funciones más usadas
print(len("Hola"))           # 4 — largo de un string
print(len([1, 2, 3, 4, 5]))  # 5 — largo de una lista

print(abs(-15))               # 15 — valor absoluto
print(round(3.14159, 2))      # 3.14 — redondear

print(max(3, 1, 4, 1, 5, 9)) # 9 — máximo
print(min(3, 1, 4, 1, 5, 9)) # 1 — mínimo

print(sum([1, 2, 3, 4, 5]))  # 15 — suma de lista

print(type(42))               # <class 'int'>
print(type("hola"))           # <class 'str'>
````

````python
# ✏️ Conversión de tipos
numero_texto = "42"
numero_entero = int(numero_texto)    # str → int
numero_decimal = float("3.14")       # str → float
texto_numero = str(100)              # int → str

print(numero_entero + 8)    # 50
print(numero_decimal * 2)   # 6.28
print("El resultado es: " + texto_numero)
````

````python
# ✏️ range() — generar secuencias de números
print(list(range(5)))         # [0, 1, 2, 3, 4]
print(list(range(1, 6)))      # [1, 2, 3, 4, 5]
print(list(range(0, 11, 2)))  # [0, 2, 4, 6, 8, 10] — de 2 en 2

# Combinado con funciones propias:
def tabla_de_multiplicar(n):
    for i in range(1, 11):
        print(f"{n} x {i} = {n * i}")

tabla_de_multiplicar(7)
````
