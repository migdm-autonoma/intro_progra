## Ejercicio definición de función

Se añade un día extra al calendario casi cada cuatro años, el 29 de febrero, y se le llama día bisiesto . Esto corrige el calendario teniendo en cuenta que nuestro planeta tarda aproximadamente 365,25 días en orbitar el sol. Un año bisiesto incluye un día bisiesto.

En el calendario gregoriano, se utilizan tres condiciones para identificar los años bisiestos:

El año es divisible por 4, es un año bisiesto, a menos que:
El año se puede dividir exactamente entre 100, NO es un año bisiesto, a menos que:
El año también es divisible exactamente por 400. Entonces es un año bisiesto.
Esto significa que, en el calendario gregoriano, los años 2000 y 2400 son bisiestos, mientras que 1800, 1900, 2100, 2200, 2300 y 2500 NO son bisiestos. Fuente

Tarea

Dado un año, determina si es un año bisiesto. Si es un año bisiesto, devuelve el valor booleano True, de lo contrario, devuelve False.

Tenga en cuenta que el fragmento de código proporcionado lee desde STDIN y pasa los argumentos a la is_leapfunción. Solo es necesario para completar la is_leapfunción.

Formato de entrada

Leer, el año para poner a prueba.

Restricciones


Formato de salida

La función debe devolver un valor booleano (Verdadero/Falso). La salida se gestiona mediante el fragmento de código proporcionado.

Entrada de ejemplo 0

1990
Salida de ejemplo 0

FALSO
Explicación 0

1990 no es múltiplo de 4, por lo tanto, no es un año bisiesto.

```python
def es_bisiesto(año):
    bisiesto = False
    
    if (año % 4 == 0) and (año % 100 != 0):
        bisiesto = True
        
    elif (año % 100 == 0) and (año % 400 != 0):
        bisiesto = False
        
    elif (año % 400 == 0):
        bisiesto = True
        
    else:
        bisiesto = False
    
    
    return bisiesto

año = int(input())
print(es_bisiesto(año))
```
